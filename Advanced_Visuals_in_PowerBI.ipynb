**1. What is the difference between a Basic Filter, Advanced Filter, and a Slicer in Power BI?**

Answer:

**Basic Filter:** Filters data by selecting specific values from a field.

**Advanced Filter:** Allows more detailed filtering using conditions such as “contains,” “is greater than,” or multiple criteria.

**Slicer:** A visual filter placed on the report page that lets users easily select values to filter other visuals.

**4. What is a Visual-Level Filter, and how does it differ from Page-Level and Report-Level Filters?**

Answer:

* **Visual-Level Filter:** Filters data for only one specific visual (chart, table, etc.).
* **Page-Level Filter:** Applies the filter to all visuals on a particular report page.
* **Report-Level Filter:** Applies the filter to all pages and visuals in the entire report.




**6. What is the role of Drill-Down in Power BI? How does it improve data exploration?**

Answer:

**Drill-Down** in Power BI allows users to move from summary-level data to more detailed information within a visual, such as from Year → Quarter → Month.

It improves data exploration by helping users analyze data at different levels of detail without creating separate reports or visuals.


**10. Explain the difference between Drill-Down and Drill-Through in Power BI using a real-world example.**

Answer:

**Drill-Down:** It is used to view data at a more detailed level within the same visual. For example, Sales by Year → Quarter → Month.

**Drill-Through:** It is used to move to another report page to see detailed information about a selected item. For example, selecting a salesperson to view their detailed sales.


**13. What is the “Keep All Filters” option in Drill-Through? How does it affect user interaction.**

Answer:

**Keep All Filters** in Power BI Drill-Through carries all the filters from the current report page to the drill-through page. This ensures that the detailed page shows data based on the same selections made by the user.

For example, if a user selects 2026 and North Region, the drill-through page will show details only for 2026 and the North Region. This makes the interaction more focused and provides consistent, filtered results.
